```python
import pandas as pd

df = pd.read_csv("../data/Telco_Customer_Churn.csv")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

#BAR CHART OF CHURN COUNTS
sns.countplot(x="Churn", data=df)
plt.title("Number of Customers: Churned vs Stayed")
plt.xlabel("Churn")
plt.ylabel("Count")
plt.show()
#Far more customers stayed than churned, 
#meaning the dataset is imbalanced this matters later when evaluating a model, 
#since accuracy alone can be misleading
```

```python
#PIE CHART OF CHURN PERCENTAGE
df["Churn"].value_counts().plot.pie(autopct="%1.1f%%", labels=["No", "Yes"], colors=["skyblue", "salmon"])
plt.title("Churn Percentage")
plt.ylabel("")
plt.show()
#About 26–27% of customers churned, which is a substantial minority worth predicting accurately
```

```python
#HISTOGRAM OF MONTHALY CHARGES
df["MonthlyCharges"].hist(bins=30, edgecolor="black")
plt.title("Distribution of Monthly Charges")
plt.xlabel("Monthly Charges")
plt.ylabel("Number of Customers")
plt.show()
#There's a cluster of customers around $20/month (likely basic plans) and another bump around $70–90/month (likely bundled plans)
```

```python
#BOX PLOT COMPARING MONTHLY CHARGES BY CHURN STATUS
sns.boxplot(x="Churn", y="MonthlyCharges", data=df)
plt.title("Monthly Charges by Churn Status")
plt.show()
#Customers who churned tend to have higher monthly charges than those who stayed, suggesting price sensitivity plays a role
```

```python
#CORRELATION HEATMAPS FOR NUMERICAL COLUMNS ONLY 
plt.figure(figsize=(8,6))
numeric_df = df.select_dtypes(include=["float64", "int64"])
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
#tenure and TotalCharges are strongly positively correlated, which makes sense, longer-tenured customers accumulate more total charges
```

```python
#CHURN BY CONTRACT TYPE (GROUPED BAR CHARTS)
sns.countplot(x="Contract", hue="Churn", data=df)
plt.title("Churn by Contract Type")
plt.show()
#Month-to-month contract customers churn far more than one- or two-year contract customers, suggesting long-term contracts reduce churn
```

```python
#CHURN BY INTERNET SERVICE TYPE (GROUPED BAR GRAPH)
sns.countplot(x="InternetService", hue="Churn", data=df)
plt.title("Churn by Internet Service Type")
plt.show()
#Customers with Fiber optic internet churn more than DSL or no-internet customers possibly due to higher pricing or service issues
```
